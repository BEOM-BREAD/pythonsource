```python
# abs(x) : 절대값 리턴

print(abs(3))
print(abs(-3))
print(abs(3.12))
```

```python
# chr() => ord()
# chr() : 유니코드 값을 넣으면 그 코드에 해당하는 문자를 리턴
# ord() : 문자를 넣으면 그 문자에 해당하는 유니코드를 리턴

print(chr(97))
print(chr(44032))

print(ord('a'))
print(ord('가'))
```

```python
# dir() : 객체가 지닌 변수나 함수를 보여주는 함수

list1 = [1,2,3,4,5]
dir(list1)

import math
dir(math)
```

```python
# divmod(a,b) : (a,b)의 (몫,나머지)로 튜플로 반환
quot, rem = divmod(7,3)
print(quot) # 몫
print(rem)  # 나머지
```

```python
# enumerate() : 리스트, 튜플, 문자열을 입력으로 받아 인덱스 값을 포함하는 enumerate 객체 반환

for i, name in enumerate(['body','foo','bar'], start=1):
    print(i, name)

for i, name in enumerate('bodybar', start=1):
    print(i, name)
```

```python
print(max([1,2,3]))
print(max("python"))
print(min([1,2,3]))
print(min("python"))
```

```python
# round(숫자 [, ndigits]) : 반올림

print(round(4.6))
print(round(4.6478, 2))
```

```python
# sorted() : 입력 데이터를 정렬 후 결과를 새로운 리스트로 리턴
list1 = [1,3,2,6,5]

# 원본 정렬
# list1.sort()

# 정렬
print(sorted(list1))

#원본
print(list1)
```

```python
# zip() :  동일한 개수로 이루어진 데이터를 묶어서 반환해주는 함수

print(list(zip([1,2,3],[4,5,6])))
print(list(zip([1,2,3],[4,5,6],[7,8,9])))
print(list(zip("abc","def")))
```

```python
# 학생  5명의 점수를 입력받아 다음 정보를 출력하는 프로그램을 작성하시오

# 출력내용
# 최고점수(max), 최저점수(min), 평균점수((sum,len)소수 둘째 자리), 점수를 오름차순으로 출력(sorted), 등수와 함께 출력(enumerate)

nums = []

for i in range(5):
    score = int(input(f"점수입력 : {nums}"))
    nums.append(score)

print(f"최고점수 : {max(nums)}점")
print(f"최저점수 : {min(nums)}점")
print(f"평균점수 : {sum(nums) / len(nums)}점")
print(f"오름차순 : {sorted(nums)}")

# 등수
sorted_desc = sorted(nums,reverse=True)
for idx, score in enumerate(sorted_desc,1):
    print(f"{idx}등 : {score}점")
```

```python
# 문자열을 입력받아 다음 정보를 출력
# 출력내용 : 입력한 문자, 각 문자의 유니코드 값, 가장 큰 문자, 가장 작은 문자

# ord, chr, max, min, list
# 문자열 입력 : Python
# P -> 80
# y -> 121

# 가장 큰 문자 : y
# 가장 작은 문자 : P

ans = input("문자 입력 : ")
print(f"입력한 문자 : {ans}")

for i in ans:
    print(f"{i} -> {ord(i)}")

print(f"가장 큰 문자 : {max(ans)}")
print(f"가장 작은 문자 : {min(ans)}")
```

```python
# 사용자로부터 숫자 10개 입력받기
nums = []

for i in range(10):
    num = int(input(f"점수입력 :"))
    nums.append(num)

# 출력
# 입력리스트, 중복제거, 오름차순, 튜플변환, 집합 크기
# ex) 3 5 2 5 7
# 원본[3,5,2,5 ...]
print(f"원본 : {nums}")
# 중복제거 {1,2,3,4}
unique_list = list(set(nums))
print(f"중복제거 : {unique_list}")
# 정렬 sorted()
sorted_nums = sorted(unique_list)
print(f"정렬 : {sorted_nums}")
# 튜플 tuple()
print(f"리스트 => 튜플 : {tuple(sorted_nums)}")
# 집합 크기 : len()
print(f"중복 제거 후 집합 크기 : {len(sorted_nums)}")
```
