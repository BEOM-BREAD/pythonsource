#### 정규표현식(정규식)
- 복잡한 문자열을 처리할 때 사용하는 방법
- 파이썬만의 고유한 문법이 아니라 문자열을 처리하는 모든 곳에서 사용하는 일종의 형식 언어

```python
# 주민등록번호 => 뒷자리를 모두  * 문자로 변경

# 파이썬에서 정규식 지원 모듈
import re

dir(re)

# re.complie() : 정규식을 작성
pattern = re.compile("ab*")

# 패턴을 특정 문자열이 포함하고 있는가?
print(pattern.match("abc"))
```

| 메서드 | 목적 |
|----|----|
| match | 문자열의 처음부터 정규식과 매치되는지 조사 |
| search | 문자열 전체를 검색하여 정규식과 매치되는지 조사 |
| findall | 정규식과 매치되는 모든 문자열(substring)을 리스트로 리턴 |
| finiter | 정규식과 매치되는 모든 문자열(substring)을 이터레이터 객체로 리턴 |

```python
# [] : [] 안에 들어있는 문자들과 매치한가

p = re.compile("[abc]")

print(p.match("a"))
print(p.match("before"))
print(p.match("dude"))
```

```python
m = p.match("a")

print(f"패턴시작 : {m.start()}")
print(f"패턴 끝 : {m.end()}")
print(f"패턴과 일치하는 문자열 반환 : {m.group()}")
print(f"일치의 시작, 끝 반환 : {m.span()}")

if m:
    print(f"Match found {m.group()}")
else:
    print("No Match")
```

- 자주 사용하는 문자 표현식

| 정규표현식 | 설명 |
|----|----|
| \d | 숫자와 매치 [0-9] 와 동일한 표현식 |
| \D | 숫자가 아닌 것과 매치 [^0-9] 와 동일한 표현식 |
| \s | 화이트페이스(스페이, 탭, 줄빠굼 등)문자와 매치 [빈칸\t\n\r\f\v]와 동일한 표현식 |
| \S | 화이트페이스(스페이, 탭, 줄빠굼 등)문자기 아닌것과 매치 [^빈칸\t\n\r\f\v]와 동일한 표현식 |
| \w | 문자 + 숫자와 매치 [a-zA-Z0-9] 와 동일한 표현식|
| \W | 문자 + 숫자가 아닌 것과 매치 [^a-zA-Z0-9] 와 동일한 표현식|

```python
p = re.compile("[0-9]+")
p = re.compile(r"\d+")
print(p.match("012 abc"))

# 원본 문자열에서 0~9가 아닌 문자 찾기
# p = re.compile("[^0-9]+")
p = re.compile(r"\D+")
print(p.match("012 abc"))

# 공백, 탭, 엔터 문자 찾기
# p = re.compile("[ \t\n\r\f\v]+")
p = re.compile(r"\s+")
print(p.search("012 abc"))

# 공백, 탭, 엔터 아닌 문자 찾기
# p = re.compile("[^ \t\n\r\f\v]+")
p = re.compile(r"\S+")
print(p.search("012 abc"))

# 대소문자, 숫자, _ 포함한 문자 찾기
# p = re.compile("[a-zA-Z0-9]+")
p = re.compile(r"\w+")
print(p.search("012 abc"))

# 대소문자, 숫자, _ 가 아닌 문자 찾기
# p = re.compile("[^a-zA-Z0-9]+")
p = re.compile(r"\W+")
print(p.search("012 abc"))
```

```python
# re.M : MULTILINE (각 줄마다 적용)

data = """python one
life is too short
python two
you need python
python three
"""

# ^ 시작하는
# python 시작하고 스페이스문자하나 영어, 숫자,_인 문자 찾기
p = re.compile(r"^python\s\w+", re.M)
print(p.findall(data))
```

```python
# () : 그루핑

# ABCABC 문자열이 반복되는지 조사

p = re.compile("[ABC]+")
m = p.search("ABCABCABC OK?")
print(m)
print(m.group())
# print(m.group(1)) => 대괄호는 불가능
```

```python
p = re.compile("(ABC)+")
m = p.search("ABCABCABC OK?")
print(m)
print(m.group())
print(m.group(1)) # 소괄호만 가능
```

```python
with open("./data/log.txt","r") as f:
    logs = f.read()

# print(logs)
logs = logs.splitlines()    # 리스트 상태
# logs

# 로그 파일에서 필요한 정보만 추출하기
# ERROR 만 출력
# INFO 개수, ERROR 개수 출력
# 날짜별 ERROR 발생횟수 출력

err_log, info = [],0
level = re.compile(r"ERROR|INFO")
for log in logs:
    if re.search(level, log):
        if "INFO" in log:
            info += 1
        else:
            err_log.append(log)

print(err_log)

# 출력예시
print("ERROR 목록")
print("-" * 25)

# ERROR 목록
# -----------------
# Database Connection Failed
# Invalid Password
daily_error = {}
for log in err_log:
    error_split = log.split("ERROR")
    print(error_split[1].strip())
    # if error_split[0] in daily_error:
    #     daily_error[error_split[0]] += 1
    # else:
    #     daily_error[error_split[0]] = 1
    daily_error[error_split[0]] = daily_error.get(error_split[0],0) + 1

# INFO : 3
# ERROR : 2
print(f"INFO : {info}")
print(f"ERROR : {len(err_log)}")

for key, value in daily_error.items():
    print(f"{key} : {value}")

# 2026-07-20 : 1
# 2026-07-21 : 1

# word = []
# with open("./ch1/data/log.txt","r", encoding="utf-8") as f:
# word.append(f)
```

```python
error, info = 0,0
pattern = re.compile(r"(?P<date>\d{4}-\d{2}-\d{2})\s(?P<level>INFO|ERROR)\s(?P<message>.+)")

for log in logs:
    m = pattern.search(log)

    if not m:
        continue

    date = m.group('date')
    level = m.group('level')
    message = m.group('message')

    if level == "INFO":
        info += 1
    elif level == "ERROR":
        error += 1
        print(message)
        daily_error[date] = daily_error.get(date,0) + 1

# 출력예시
print("ERROR 목록")
print("-" * 25)

print(f"INFO : {info}")
print(f"ERROR : {len(err_log)}")

for key, value in daily_error.items():
    print(f"{key} : {value}")
```

```python
# 전방탐색

url = "http://google.com"

# 프로토콜 검색 ftp://~~

p = re.compile(".+:")
m = p.search(url)
print(m)
print(m.group())
# : 패턴으로는 쓰고 매칭되는 문자열 돌려받을 때는 빼고 받고 싶다면?
```

```python
# ?= : 긍정형 전방탐색
# ?! : 부정형 전방탐색
p = re.compile(".+(?=:)")
m = p.search(url)
print(m.group())
```
