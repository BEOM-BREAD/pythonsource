# 카테고리별 요약 정보 조회기

- 딕셔너리를 활용해 "선택한 항목에 맞는 정보"를 찾아 보여주는 함수 구현
- 여러 종류의 출력 컴포넌트(텍스트 + 숫자)를 함께 배치

### 💡 강의 포인트
- 지금은 정적인 딕셔너리 데이터를 쓰지만, 구조 자체는 이후 Oracle DB 조회 결과를 보여주는 방식과 완전히 동일합니다. (ex10에서 실제 DB로 바꿔봅니다)

```python
import gradio as gr

# Chapter02에서 다뤘던 카테고리별 통계를 미리 요약해둔 딕셔너리
# (실제로는 DB 쿼리 결과일 수도 있음)
category_summary = {
    "분식": {"평균가격": 8500,  "평균평점": 4.3, "인기메뉴": "떡볶이"},
    "치킨": {"평균가격": 22000, "평균평점": 4.7, "인기메뉴": "양념치킨"},
    "피자": {"평균가격": 25500, "평균평점": 4.2, "인기메뉴": "포테이토피자"},
    "중식": {"평균가격": 10000, "평균평점": 4.3, "인기메뉴": "마라탕"},
    "일식": {"평균가격": 12000, "평균평점": 4.4, "인기메뉴": "초밥세트"},
}

def show_summary(category):
    info = category_summary[category]
    # 딕셔너리 안의 값들을 각각 꺼내서 반환 (outputs 3개와 순서를 맞춤)
    return info["평균가격"], info["평균평점"], info["인기메뉴"]
```

```python
demo = gr.Interface(
    fn=show_summary,
    inputs=gr.Dropdown(list(category_summary.keys()), label="카테고리 선택"),
    outputs=[
        gr.Number(label="평균 가격(원)"),
        gr.Number(label="평균 평점"),
        gr.Textbox(label="인기 메뉴"),
    ],
    title="카테고리별 요약 조회기",
)
demo.launch()
```
