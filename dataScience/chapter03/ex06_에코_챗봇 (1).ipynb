# ChatInterface + 대화 횟수 세기
- `gr.ChatInterface`로 대화형 채팅 화면
- `history` 매개변수를 실제로 활용해서 "지금까지 몇 번 대화했는지" 세는 기능을 추가

### 💡 강의 포인트
- 지금은 "사용자가 입력한 말을 그대로 따라 하는(echo)" 챗봇이지만, 함수 안의 로직만 AI 모델 호출로 바꾸면 진짜 챗봇이 됩니다.
- `history`는 지금까지 오간 대화 기록이 리스트로 자동 전달됩니다. **현재 메시지는 포함되지 않은** "이전까지의 기록"이라는 점이 핵심입니다.
- 최신 Gradio는 `history`를 `{"role": "user"/"assistant", "content": ...}` 형태의 딕셔너리 리스트로 줍니다. 대화 1턴(주고받음 1번)마다 **user 1개 + assistant 1개, 총 2개 항목**이 쌓이기 때문에, 실제 "대화 턴 수"를 구하려면 `len(history) // 2`로 나눠줘야 합니다.

```python
import gradio as gr

# ChatInterface에 넘기는 함수는 (message, history) 두 개를 받음
#   message : 사용자가 방금 입력한 메시지
#   history : 지금까지의 대화 기록 (현재 메시지는 포함 안 됨)
def echo_bot(message, history):
    # history는 [{"role": "user", "content": ...}, {"role": "assistant", "content": ...}, ...]
    # 형태의 리스트입니다. 대화 1턴마다 user+assistant 2개 항목이 쌓임
    #
    # 예시)
    #   대화를 한 번도 안 했을 때   -> history = []                (길이 0)
    #   1번 주고받았을 때           -> history = [user, assistant]  (길이 2)
    #   2번 주고받았을 때           -> history = [user, assistant, user, assistant]  (길이 4)
    #
    # 그래서 "지금까지 몇 번 주고받았는지"는 len(history) // 2 로 구하고,
    # 지금 들어온 메시지는 "새로운 한 턴"이므로 마지막에 +1을 해줌
    turn_count = len(history) // 2 + 1

    return f"[{turn_count}번째 대화] 너가 말한 건: '{message}' 이지?"
```

```python
demo = gr.ChatInterface(
    fn=echo_bot,
    title="대화 횟수를 세는 에코 챗봇",
    description="입력한 말을 그대로 따라 하면서, 지금이 몇 번째 대화인지도 함께 알려줍니다.",
)
demo.launch()
```
