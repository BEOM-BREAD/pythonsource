# Hello Gradio — 첫 인터페이스 만들기
- `gr.Interface`로 가장 기본적인 웹 인터페이스를 만드는 흐름
- "함수 하나 → 입력창 하나 → 출력창 하나"라는 Gradio의 기본 구조를 이해

### 💡 강의 포인트
- Gradio의 핵심은 **"평범한 파이썬 함수를 웹 화면으로 자동 변환해준다"**는 것. 함수만 잘 만들면 나머지(HTML, 버튼, 입력창)는 Gradio가 알아서 그려줍니다.
- `demo.launch()`를 실행하면 로컬 서버가 켜지고, 노트북 안에 화면이 바로 임베드됩니다(`inline=True`가 기본값).

```python
# gradio 라이브러리를 gr이라는 별명으로 불러옵니다.
import gradio as gr

# ─────────────────────────────────────────────
# 1단계 : 평범한 파이썬 함수를 하나 만듭니다.
#         이 함수가 Gradio 화면의 "핵심 로직"이 됩니다.
# ─────────────────────────────────────────────
def greet(name):
    # 입력받은 이름(name)을 이용해 인사말 문자열을 만들어 반환
    return f"안녕하세요, {name}님! 반갑습니다."


# ─────────────────────────────────────────────
# 2단계 : gr.Interface로 이 함수를 웹 화면으로 감싸줍니다.
#   - fn      : 실행할 함수
#   - inputs  : 입력 컴포넌트 종류 ("text" = 한 줄 텍스트 입력창)
#   - outputs : 출력 컴포넌트 종류 ("text" = 텍스트 출력창)
# ─────────────────────────────────────────────
demo = gr.Interface(
    fn=greet,
    inputs="text",
    outputs="text",
    title="첫 Gradio 인터페이스",          # 화면 상단 제목
    description="이름을 입력하면 인사말을 보여줍니다.",  # 화면 상단 설명
)
```

```python
# 3단계 : 실제로 서버를 켜서 화면을 띄웁니다.
# inline=True(기본값)라서 노트북 셀 안에 바로 화면이 나타납니다.
demo.launch()
```

```python

# ⚠️ 실습 팁: 노트북에서 Gradio를 계속 켜두면 포트가 쌓여서 다음 셀 실행이 꼬일 수 있습니다.
# 확인이 끝나면 아래처럼 데모를 꺼주는 습관을 들이세요.
demo.close()
```
