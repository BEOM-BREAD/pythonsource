# 이미지 업로드 + 흑백 변환기
- `gr.Image`로 이미지를 업로드받고 처리 결과를 다시 이미지로 보여주는 출력
- 텍스트/숫자뿐 아니라 이미지도 Gradio의 입출력이 가능

### 💡 강의 포인트
- Gradio는 이미지를 기본적으로 **numpy 배열**로 받아옵니다. 그래서 PIL, OpenCV 같은 이미지 라이브러리와 자연스럽게 연동됩니다.
- 흑백 변환처럼 간단한 처리부터 시작하면, 나중에 "AI 모델에 이미지를 넣어서 결과를 보여주는" 응용으로 확장하기 쉽습니다.

```python
# 이미지 배열 계산을 위해 numpy를 사용합니다.
import gradio as gr
import numpy as np

def to_grayscale(input_image):
    # input_image는 (세로, 가로, 3) 모양의 numpy 배열로 들어옵니다. (3 = R,G,B)
    # 컬러 사진을 흑백으로 바꾸는 공식 : R,G,B에 가중치를 곱해서 합칩니다.
    # (사람 눈이 초록색에 더 민감하다는 특성을 반영한 표준 공식입니다)
    gray = np.dot(input_image[..., :3], [0.299, 0.587, 0.114])

    # 흑백 값을 다시 (세로, 가로, 3) 모양으로 복원해서 회색조 이미지처럼 보이게 함
    gray_image = np.stack([gray, gray, gray], axis=-1).astype(np.uint8)
    return gray_image
```

```python
demo = gr.Interface(
    fn=to_grayscale,
    inputs=gr.Image(label="원본 이미지 업로드"),
    outputs=gr.Image(label="흑백 변환 결과"),
    title="이미지 흑백 변환기",
)
demo.launch()
```
